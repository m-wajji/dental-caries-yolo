```python
# check your setup
import torch, ultralytics
print("PyTorch:", torch.__version__)
print("GPU available:", torch.cuda.is_available())
ultralytics.checks()
```

```python
# run a pretrained detector

from ultralytics import YOLO

model = YOLO("yolo26n.pt")     # YOLO26 nano: latest Ultralytics model, smallest size
results = model("https://ultralytics.com/images/bus.jpg")
results[0].show()
```

```python
# look inside a prediction

r = results[0]
print(r.boxes.xyxy)    # box corners in pixels: x1, y1, x2, y2
print(r.boxes.conf)    # confidence per box
print(r.boxes.cls)     # class ID per box
print(r.names)         # class ID -> name
```
